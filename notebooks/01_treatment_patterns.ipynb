# 01 — Annual Treatment Patterns

Demonstrates the original project's descriptive question: how CIT use changed across diagnosis years. The data are synthetic and do not reproduce NCDB results.

```
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path("..")
df = pd.read_csv(ROOT / "data" / "synthetic_dlbcl_demo.csv")
df.head()
```

```
annual = (
    df.assign(CIT=(df["treatment"]=="Chemoimmunotherapy").astype(int))
      .groupby("diagnosis_year")["CIT"]
      .agg(["mean","count"])
      .reset_index()
)
annual["CIT_percent"] = 100*annual["mean"]
annual
```

```
plt.figure(figsize=(7,4))
plt.plot(annual["diagnosis_year"], annual["CIT_percent"], marker="o")
plt.xlabel("Diagnosis year")
plt.ylabel("Chemoimmunotherapy use (%)")
plt.title("Synthetic demonstration: annual treatment distribution")
plt.show()
```
