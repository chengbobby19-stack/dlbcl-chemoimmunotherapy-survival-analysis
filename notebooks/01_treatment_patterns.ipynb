# 01 — Annual Treatment Patterns

Synthetic demonstration of the original project's descriptive question: how chemoimmunotherapy (CIT) use changed across diagnosis years.

```
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path("..")
df = pd.read_csv(ROOT / "data" / "synthetic_dlbcl_demo.csv")
df.head()
```

```
annual = (
    df.assign(CIT=(df["treatment"]=="Chemoimmunotherapy").astype(int))
      .groupby("diagnosis_year")["CIT"]
      .agg(["mean","count"])
      .reset_index()
)
annual["CIT_percent"] = 100*annual["mean"]
annual["CT_percent"] = 100-annual["CIT_percent"]
annual
```

```
plt.figure(figsize=(7,4))
plt.plot(annual["diagnosis_year"], annual["CIT_percent"], marker="o", label="CIT")
plt.plot(annual["diagnosis_year"], annual["CT_percent"], marker="o", label="CT alone")
plt.xlabel("Year of diagnosis")
plt.ylabel("Percentage")
plt.title("Synthetic demonstration — Annual treatment distribution")
plt.legend()
plt.show()
```
