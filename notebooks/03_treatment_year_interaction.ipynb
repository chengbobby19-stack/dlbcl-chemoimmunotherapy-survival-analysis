# 03 — Treatment-by-Year Interaction

Synthetic demonstration of the treatment-by-calendar-year interaction used to assess whether the relative treatment association changes over time.

```
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path("..")
df = pd.read_csv(ROOT / "data" / "synthetic_dlbcl_demo.csv")
df.head()
```

```
from statsmodels.duration.hazard_regression import PHReg

m = df.copy()
m["CIT"] = (m["treatment"]=="Chemoimmunotherapy").astype(int)
m["year_centered"] = m["diagnosis_year"] - m["diagnosis_year"].min()
m["CIT_x_year"] = m["CIT"] * m["year_centered"]
m["male"] = (m["sex"]=="Male").astype(int)
m["stage_III"] = (m["ann_arbor_stage"]=="III").astype(int)
m["stage_IV"] = (m["ann_arbor_stage"]=="IV").astype(int)

X = m[["CIT","year_centered","CIT_x_year","age","male","stage_III","stage_IV","comorbidity_ge1","b_symptoms"]].astype(float)
model = PHReg(m["survival_months"], X, status=m["death"])
result = model.fit()
print(result.summary())
```
