# 03 — Treatment-by-Year Interaction

The original study assessed whether the relative treatment-survival association changed across calendar time. This notebook demonstrates that model structure on synthetic data.

```
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path("..")
df = pd.read_csv(ROOT / "data" / "synthetic_dlbcl_demo.csv")
df.head()
```

```
from statsmodels.duration.hazard_regression import PHReg

m = df.copy()
m["CIT"] = (m["treatment"]=="Chemoimmunotherapy").astype(int)
m["year_centered"] = m["diagnosis_year"] - 2013
m["CIT_x_year"] = m["CIT"] * m["year_centered"]
m["male"] = (m["sex"]=="Male").astype(int)
m["advanced_stage"] = (m["stage"]=="III-IV").astype(int)

X = m[["CIT","year_centered","CIT_x_year","age","male","advanced_stage","comorbidity_ge1","b_symptoms"]].astype(float)
model = PHReg(m["survival_months"], X, status=m["death"])
result = model.fit()
print(result.summary())
```
